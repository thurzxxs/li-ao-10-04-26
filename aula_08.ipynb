```python
# escrever um programa que verifica se o número digitado é par ou impar

# entrada

numero = int (input("Digite um número: "))

# processamento / saída

if numero % 2 == 0:

  print (f" O número {numero} é par.")

else:
  print (f"O número {numero} é impar")
```

```python
# escrever um programa que leia o valor de uma compra e aplique:

# Até R$ 100 - sem desconto
# R$ 101 - a R$ 500 - 10 % de desconto
# Acima de R$ de 500 - desconto de 20 %

# mostrar valor a pagar depois de aplicativo o desconto

# entrada

compra = float (input ("digite o valor da sua compra:"))

desconto = compra - (compra * 0.20)

if compra > 100:
  print(f"o valor a pagar é: {desconto}")

elif compra > 500:
  compra = compra - (compra * 0.20)
  print (f"o valor a pagar é: {desconto}")

else:
  print (f"você não obteve desconto, valor a pagar: {compra} ")
```
